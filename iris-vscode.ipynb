```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import ConfusionMatrixDisplay


models = [
    ('LR',  LogisticRegression(solver='lbfgs', max_iter=1000)),  # multinomial by default
    ('LDA', LinearDiscriminantAnalysis()),
    ('KNN', KNeighborsClassifier()),
    ('CART', DecisionTreeClassifier(random_state=1)),
    ('NB',  GaussianNB()),
    # SVM + scaling is a good practice
    ('SVM', make_pipeline(StandardScaler(), SVC(gamma='auto')))
]
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

results, names = [], []
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=1)
for name, model in models:
    cv = cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy')
    results.append(cv); names.append(name)
    print(f"{name}: {cv.mean():.6f} ({cv.std():.6f})")
```

```python
# Visualize model comparison
import matplotlib.pyplot as plt
plt.boxplot(results, labels=names)
plt.title("Algorithm Comparison (10-fold CV)")
plt.show()
```

```python
# Feature/target split
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# If 'dataset' is your DataFrame from the CSV:
X = dataset[['sepal-length','sepal-width','petal-length','petal-width']].to_numpy()
y = dataset['class'].to_numpy()

# Train/test split (stratified to keep class balance)
X_train, X_test, Y_train, Y_test = train_test_split(
    X, y, test_size=0.20, random_state=1, stratify=y
)

# quick sanity checks
print(X_train.shape, X_test.shape)
print(pd.Series(Y_train).value_counts(), "\n", pd.Series(Y_test).value_counts())
```

```python
best = make_pipeline(StandardScaler(), SVC(gamma='auto'))
best.fit(X_train, Y_train)
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
y_pred = best.predict(X_test)
print("Test accuracy:", accuracy_score(Y_test, y_pred))
print(confusion_matrix(Y_test, y_pred))
print(classification_report(Y_test, y_pred))
```

```python
X = dataset[['sepal-length','sepal-width','petal-length','petal-width']].to_numpy()
y = dataset['class'].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=1, stratify=y
)

X_train.shape, X_test.shape, pd.Series(y_train).value_counts(), pd.Series(y_test).value_counts()
```

```python
# Standard practice: scale features for SVM
model = make_pipeline(
    StandardScaler(),
    SVC(kernel="rbf", gamma="scale", C=1.0, random_state=1)  # default SVC; probability=False keeps it fast
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification report:\n", classification_report(y_test, y_pred))

# Confusion matrix (nice visual)
disp = ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.title("SVM on Iris — Confusion Matrix")
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

pipe = make_pipeline(StandardScaler(), SVC(random_state=1))
param_grid = {
    "svc__C": [0.1, 1, 10, 100],
    "svc__gamma": ["scale", 0.1, 0.01, 0.001],
    "svc__kernel": ["rbf"]  # keep rbf for this dataset
}
grid = GridSearchCV(pipe, param_grid, cv=10, scoring="accuracy", n_jobs=-1)
grid.fit(X_train, y_train)

print("Best params:", grid.best_params_)
print("CV best score:", grid.best_score_)
best_model = grid.best_estimator_
print("Test accuracy:", accuracy_score(y_test, best_model.predict(X_test)))
```
